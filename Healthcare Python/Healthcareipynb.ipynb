```python
temp=df1.merge(df[['patient_id','date_of_birth']],on='patient_id',how='left')
```

```python
df1['Age_at_admission'] = ((temp['admission_date'] - temp['date_of_birth']).dt.days/365.25).astype('int64')
```

```python
df1['Age_group']=pd.cut(df1['Age_at_admission'],bins=[0,18,35,50,65,120],labels=['0-18','19-35','36-50','51-65','65+']
)
```

```python
df1.head()
```

```python
df1=df1.rename(columns={'Age_group':'Age_band'})
```

```python
df1['Age_Group'] = pd.cut(
    df1['Age_at_admission'],
    bins=[0, 17, 35, 55, 120],
    labels=['Child', 'Young Adult', 'Adult', 'Senior Citizen']
)
```

```python
df1.head()
```

```python
df1['length_of_stay']=(df1['discharge_date']-df1['admission_date']).dt.days
```

```python
df1['stay_category']=pd.cut(df1['length_of_stay'],bins=[0,3,7,14,float('inf')],labels=['Short','Medium','Long','Extended long'])
```

```python
df2['Insurance_Coverage_Pct'] = (
    df2['insurance_covered_amount'] / df2['total_amount'] * 100
).round(2)
```

```python
df2['Insurance_Coverage_Pct'].value_counts()
```

```python
df2['Patient_Payment_Pct'] = (
    df2['patient_payable_amount'] / df2['total_amount'] * 100
).round(2)
```

```python
df2.info()
```

```python
def coverage_category(x):
  if x==100:
    return 'Fully Insured'
  elif x==0:
    return 'Self Pay'
  else:
    return 'Partially Insured'
```

```python
df2['coverage_category']=(df2['Insurance_Coverage_Pct'].apply(coverage_category))
```

```python
df2.head()
```

```python
df2['Bill_Category'] = pd.cut(
    df2['total_amount'],
    bins=[0,5000,15000,30000,float('inf')],
    labels=['Low','Medium','High','Very High']
)
```
